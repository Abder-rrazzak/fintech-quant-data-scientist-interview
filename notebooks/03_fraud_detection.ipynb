# Fraud score to decision

A rare label needs PR-AUC and a cost threshold, then an approve / review / block rule.

Synthetic laboratory. Not a claim of production ownership.

```
import sys
from pathlib import Path
sys.path.insert(0, str(Path('..').resolve() / 'src'))
```

```
import numpy as np
from fraud.decisions import decide
from evaluation.thresholds import select_threshold, threshold_cost_curve
y = np.array([1, 0, 0, 1, 0])
scores = np.array([0.9, 0.6, 0.2, 0.4, 0.05])
curve = threshold_cost_curve(y, scores, fn_cost=500, fp_cost=5)
best = select_threshold(curve)
decide(scores, best['threshold'], min(0.99, best['threshold'] + 0.2))
```
